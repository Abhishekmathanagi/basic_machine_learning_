```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
x,y_true=make_blobs(n_samples=500,centers=3,cluster_std=1.6,random_state=42)
```

```python
df=pd.DataFrame(x,columns=['feature1','feature2'])
```

```python
df
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(x)
```

```python
ineria=[]
k_range=range(1,11)
```

```python
for k in k_range:
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit(X_scaled)
    ineria.append(kmeans.inertia_)
```

```python
ineria
```

```python
plt.plot(k_range,ineria,marker='o')
```

```python
final_Kmeans=KMeans(n_clusters=3,random_state=42)
```

```python
cluster_labels=final_Kmeans.fit_predict(X_scaled)
```

```python
df['cluster']=cluster_labels
```

```python
cluster_labels
```

```python
sns.scatterplot(x=df['feature1'],
                y=df['feature2'],
                hue=df['cluster'],
                palette='viridis'

)
```

```python
from sklearn.datasets import make_moons
```

```python
X,y_true=make_moons(n_samples=500,noise=0.5,random_state=42)
```

```python
from sklearn.cluster import KMeans,DBSCAN
```

```python
df=pd.DataFrame(X,columns=['feature1','feature2'])
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(df)
```

```python
kmeans=KMeans(n_clusters=2,random_state=42)
kmeans_labels=kmeans.fit_predict(x_scaled)
```

```python
df['kmeans_cluster']=kmeans_labels
```

```python
sns.scatterplot(x=df['feature1'],
                y=df['feature2'],
                hue=df['kmeans_cluster'],
                palette='tab10')
```

```python
dbscan=DBSCAN(eps=0.3,min_samples=5)
dbscan_labels=dbscan.fit_predict(x_scaled)
```

```python
df['dbscan_cluster']=dbscan_labels
```

```python
sns.scatterplot(x=df['feature1'],
                y=df['feature2'],
                hue=df['dbscan_cluster'],
                palette='tab10')
```
