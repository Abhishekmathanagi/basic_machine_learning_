```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('train.txt',sep=';', header=None,names=['text','emotion'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotion=df['emotion'].unique()
```

```python
emotion_num={}
i=0
for emo in unique_emotion:
  emotion_num[emo]=i
  i+=1

df['emotion']=df['emotion'].map(emotion_num)
```

```python
df
```

```python
df['text']=df['text'].apply(lambda x:x.lower())
```

```python
import string
def remove_punc(text):
  return text.translate(str.maketrans(' ',' ',string.punctuation))
```

```python
df
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
def remove_num(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_num)
```

```python
df
```

```python
import re

def remove_url(txt):
    new = re.sub(r'https?://\S+|www\.\S+', '', txt)
    return new

df['text'] = df['text'].apply(remove_url)
```

```python
def remove_html_tags(txt):
    new = re.sub(r'<.*?>', '', txt)
    return new

df['text'] = df['text'].apply(remove_html_tags)
```

```python
def remove_emoji(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

df['text'] = df['text'].apply(remove_emoji)
```

```python
df.loc[1]['text']
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words=set(stopwords.words('english'))
```

```python
def remove(txt):
  word=word_tokenize(txt)
  new=[]
  for w in word:
    if w not in stop_words:
      new.append(w)
  return ' '.join(new)
```

```python
df['text']=df['text'].apply(remove)
```

```python
from sklearn.model_selection import train_test_split
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model=MultinomialNB()
```

```python
nb_model.fit(X_train_bow,y_train)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)


nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)


pred_bow = nb_model.predict(X_test_bow)
print(accuracy_score(y_test, pred_bow))
```
